```python
import os
import sys
import speech_recognition as sr
from contextlib import contextmanager
import webbrowser
from thefuzz import fuzz, process
from gtts import gTTS
from pygame import mixer

def fale(text):
    tts = gTTS(text=text, lang='en')
    filename = "voice.mp3"
    try:
        os.remove(filename)
    except OSError:
        pass
    tts.save(filename)
    mixer.init()
    mixer.music.load(filename)
    mixer.music.play()
    while mixer.music.get_busy():
        continue      

def ouvir_microfone():
    microfone = sr.Recognizer()

    microfone.pause_threshold = 2.0  # Dá mais tempo para você falar frases longas
    
    # Tentamos abrir o microfone e capturar áudio
    with ignore_stderr():
        try:
            # Verifica se há microfones disponíveis
            if not sr.Microphone.list_microphone_names():
                fale("Erro: Nenhum microfone detectado no sistema.")
                return None
                
            source = sr.Microphone()
            with source as s:
                fale("Ajustando ruído ambiente (aguarde 1s)...")
                microfone.adjust_for_ambient_noise(s, duration=1)
                fale("Diga alguma coisa: ") 
                # Adicionamos um timeout para não ficar preso se nada for dito
                # phrase_time_limit limita o tempo da fala capturada
                audio = microfone.listen(s, timeout=5, phrase_time_limit=100)
                return audio
        except sr.WaitTimeoutError:
            fale("Tempo esgotado: Nenhuma fala foi detectada.")
            return None
        except Exception as e:
            # Fora do contexto, o erro volta a aparecer se algo der muito errado
            # Comum em Linux: erros de permissão ou falta do PyAudio
            fale(f"Erro ao acessar o microfone: {e}")
            if "pyaudio" in str(e).lower():
                fale("Dica: Certifique-se que o 'pyaudio' está instalado.")
            return None

def descreve_audio(audio):
    if audio is None:
        return "Sem áudio para processar."
        
    recognizer = sr.Recognizer()
    try:
        print("Traduzindo fala em texto...")
        frase = recognizer.recognize_google(audio, language='pt-BR')
        return frase
    except sr.UnknownValueError:
        return "Não entendi o que foi dito."
    except sr.RequestError as e:
        return f"Erro no serviço: {e}"

def executar_acao(texto):
    """
    Analisa o texto usando NLP (fuzzy matching) e executa a ação correspondente.
    """
    if texto is None or "Não entendi" in texto or "Erro" in texto:
        return "Comando inválido ou não reconhecido."

    comando = texto.lower()
    
    # Definição de intenções e alvos usando sinonimos
    verbos_abrir = ["abrir", "abre", "entra", "acessa", "vai para", "mostrar", "quero ver"]
    alvos_facebook = ["facebook", "face", "rede social", "meu perfil"]

    # Verifica se há uma correspondência fuzzy para o verbo de abertura
    best_verbo, score_verbo = process.extractOne(comando, verbos_abrir, scorer=fuzz.partial_ratio)
    
    # Verifica se há uma correspondência fuzzy para o alvo Facebook
    best_alvo, score_alvo = process.extractOne(comando, alvos_facebook, scorer=fuzz.partial_ratio)

    print(f"DEBUG NLP: Verbo '{best_verbo}' ({score_verbo}%), Alvo '{best_alvo}' ({score_alvo}%)")

    # Se ambos tiverem um score alto (ajustável), executamos a ação
    if score_verbo > 70 and score_alvo > 70:
        try:
            fale(f"Interpretado como: '{best_verbo} {best_alvo}' -> Abrindo Facebook...")
            url = "https://www.facebook.com"
            webbrowser.open(url)
            return f"Ação 'Abrir Facebook' executada (Entendido: {best_verbo} {best_alvo})."
        except Exception as e:
            return f"Erro ao tentar abrir o navegador: {e}"
    
    return f"Comando '{texto}' reconhecido, mas não entendi qual ação realizar."

if __name__ == "__main__":
    audio_capturado = ouvir_microfone()
    if audio_capturado:
        resultado_texto = descreve_audio(audio_capturado)
        fale("\nTexto reconhecido: " + resultado_texto)
        
        # Em vez de apenas dar print, executamos a ação
        status_acao = executar_acao(resultado_texto)
        fale("Status da Ação: " + status_acao)
```
